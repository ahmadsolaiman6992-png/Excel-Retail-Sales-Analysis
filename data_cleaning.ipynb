# Data Cleaning - Retail Sales (Excel Project)

Cleans the `fact_sales` sheet before analysis:
1. Remove fully empty rows
2. Standardize product codes (`StockCode`): trim whitespace + uppercase
3. Save the cleaned data to CSV

**Note:** Power Query was not available in my environment, so the cleaning was done with Pandas.
Place `all_fact_sales2026.xlsx` in the same folder as this notebook before running.

## 1. Load data

```python
import pandas as pd

# Original workbook must be in the same folder as this notebook
df = pd.read_excel("all_fact_sales2026.xlsx", sheet_name="fact_sales")
len(df)
```

## 2. Remove fully empty rows

```python
df = df.dropna(how="all")
df.shape  # same row count as before -> no fully empty rows found
```

## 3. Standardize product codes (StockCode)

```python
# Unique codes BEFORE standardizing
df["StockCode"].nunique()
```

```python
# Trim whitespace and convert to uppercase (e.g. "15056bl" and "15056BL" become one code)
df["StockCode"] = df["StockCode"].astype(str).str.strip().str.upper()
```

```python
# Unique codes AFTER standardizing (3925 -> 3816 = 109 variants merged)
df["StockCode"].nunique()
```

## 4. Save cleaned data

```python
df.to_csv("cleaned_sales.csv", index=False)
```
